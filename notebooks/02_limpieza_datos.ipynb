```python
inflacion_final = inflacion_ecuador.T

inflacion_final.head()
```

LIMPIEZA DE DESEMPLEO

```python
desempleo = pd.read_csv(
    "../data/raw/API_SL.UEM.TOTL.ZS_DS2_es_csv_v2_7203.csv",
    skiprows=4
)

print("Desempleo cargado correctamente")
```

```python
desempleo_ecuador = desempleo[desempleo["Country Code"] == "ECU"]

desempleo_ecuador
```

```python
desempleo_ecuador = desempleo_ecuador.drop(
    columns=["Country Name", "Country Code", "Indicator Name", "Indicator Code", "Unnamed: 70"],
    errors="ignore"
)

desempleo_ecuador
```

```python
desempleo_final = desempleo_ecuador.T

desempleo_final.head()
```

```python
desempleo_final = desempleo_final.reset_index()

desempleo_final.columns = ["Año", "Desempleo"]

desempleo_final.head()
```

```python
pib_final = pib_final.reset_index()
pib_final.columns = ["Año", "PIB"]

inflacion_final = inflacion_final.reset_index()
inflacion_final.columns = ["Año", "Inflacion"]

datos_final = pib_final.merge(
    inflacion_final,
    on="Año",
    how="inner"
)

datos_final.head()
```

```python
datos_final = datos_final.merge(
    desempleo_final,
    on="Año",
    how="inner"
)

datos_final.head()
```

PREPARAR LAS VARIABLES

```python
datos_final = datos_final.drop(
    columns=["Tasa_Interes"],
    errors="ignore"
)

datos_final.head()
```

```python
datos_final = datos_final.sort_values("Año")

datos_final
```

UNIR LAS BASES

```python
datos_final["Año"] = pd.to_numeric(datos_final["Año"], errors="coerce")
```

```python
datos_final.dtypes
```

```python
datos_final = datos_final[
    (datos_final["Año"] >= 2007) &
    (datos_final["Año"] <= 2024)
]

datos_final
```

```python
datos_final.to_csv(
    "../data/processed/base_VAR_Ecuador.csv",
    index=False
)

print("Base econométrica guardada correctamente")
```
