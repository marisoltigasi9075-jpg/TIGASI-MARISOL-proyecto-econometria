```python
import pandas as pd
from statsmodels.tsa.stattools import adfuller
```

```python
df = pd.read_csv("../data/processed/base_VAR_Ecuador.csv")

df.head()
```

```python
def prueba_adf(serie, nombre):
    resultado = adfuller(serie.dropna())

    print("=" * 50)
    print(f"Prueba Dickey-Fuller: {nombre}")
    print("=" * 50)

    print(f"ADF Statistic : {resultado[0]:.4f}")
    print(f"p-value       : {resultado[1]:.4f}")
    print(f"Número de rezagos : {resultado[2]}")
    print(f"Observaciones : {resultado[3]}")

    print("\nValores críticos:")
    for clave, valor in resultado[4].items():
        print(f"{clave}: {valor:.4f}")

    if resultado[1] < 0.05:
        print("\n✅ La serie es estacionaria.")
    else:
        print("\n❌ La serie NO es estacionaria.")
```

```python
prueba_adf(df["PIB"], "PIB")
```

```python
prueba_adf(df["Inflacion"], "Inflación")
```

```python
prueba_adf(df["Inflacion"], "Inflación")
```

PIB, INFLACION Y DESEMPLEO

```python
import pandas as pd
from statsmodels.tsa.stattools import adfuller

variables = {
    "PIB": df["PIB"],
    "Inflacion": df["Inflacion"],
    "Desempleo": df["Desempleo"]
}

resultados = []

for nombre, serie in variables.items():

    adf = adfuller(serie.dropna())

    resultados.append({
        "Variable": nombre,
        "ADF": round(adf[0],3),
        "p-value": round(adf[1],4),
        "Rezagos": adf[2],
        "Observaciones": adf[3],
        "Estacionaria": "Sí" if adf[1] < 0.05 else "No"
    })

tabla_adf = pd.DataFrame(resultados)

tabla_adf
```
